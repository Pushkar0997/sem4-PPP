```python
data = {
    'Name': ['Amit', 'Riya', 'John', None, 'Sara', 'Raj', None],
    'Age': [23, np.nan, 25, 30, None, 28, 22],
    'Marks': [85, 90, None, 70, 88, None, 95],
    'City': ['Pune', 'Mumbai', None, 'Delhi', 'Pune', None, 'Mumbai']
}
```

```python
df= pd.DataFrame(data)
```

```python
df.info()
```

```python
df.describe
```

```python
df_mean = df.copy()
```

```python
df_mean['Age'].fillna(df['Age'].mean(), inplace=True)
```

```python
df_mean
```

```python
df_mean['Marks'].fillna(df['Marks'].median())
```

```python
df_mean['Marks'].fillna(df['Marks'].median(), inplace=True)
df_mean
```

```python
df_mean['City'].fillna(df['City'].mode()[0], inplace=True)
df_mean
```

```python
df_mean['Name'].drop_duplicates
```

```python
df_copy = df.copy()
```

```python
df_copy['Age'].fillna(df_copy['Age'].mean(), inplace=True)
df_copy['Marks'].fillna(df_copy['Marks'].median(), inplace=True)
```

```python
df_copy
```

```python
df_copy.fillna(method='ffill')
```

```python
df_copy.fillna(method='bfill')
```
